1. What is a Support Vector Machine (SVM)?

A Support Vector Machine is a supervised machine learning algorithm used for both classification and regression tasks. It works by finding an optimal boundary called a hyperplane that separates different classes of data while maximizing the margin between them. The margin is the distance between the closest data points of each class and the hyperplane.

2. What is the difference between Hard Margin and Soft Margin SVM?

The difference between hard margin and soft margin SVM lies in how strictly they separate data. Hard margin SVM assumes that the data is perfectly separable and does not allow any misclassification, making it sensitive to noise. Soft margin SVM, on the other hand, allows some misclassification by introducing a penalty term, making it more practical for real world datasets.

3. What is the mathematical intuition behind SVM

The mathematical intuition behind SVM is to find a hyperplane that maximizes the margin between classes while ensuring correct classification as much as possible. It balances between minimizing classification error and maximizing the distance between the decision boundary and the nearest points.

4. What is the role of Lagrange Multipliers in SVM

Lagrange multipliers are used in SVM to transform the constrained optimization problem into a form that is easier to solve. They allow the model to focus only on important data points and express the solution in terms of dot products, which is essential for using kernel functions.

5.  What are Support Vectors in SVM?

Support vectors are the data points that lie closest to the decision boundary. These points are crucial because they directly influence the position and orientation of the hyperplane. Removing them would change the model significantly.

6. What is a Support Vector Classifier (SVC)?

A Support Vector Classifier is the application of SVM for classification problems. It determines the best boundary that separates different classes and can handle both linear and non linear classification using kernel functions.

7.  What is a Support Vector Regressor (SVR)?

A Support Vector Regressor is the version of SVM used for regression tasks. Instead of finding a boundary, it tries to fit a function that predicts continuous values while allowing a small margin of error within which predictions are considered acceptable.

8. What is the Kernel Trick in SVM?

The kernel trick is a technique that allows SVM to work efficiently with non linear data by mapping it into a higher dimensional space. Instead of explicitly transforming the data, it uses kernel functions to compute similarities directly.

9. Compare Linear Kernel, Polynomial Kernel, and RBF Kernel

The linear kernel is simple and works well when the data is linearly separable. The polynomial kernel can model interactions between features and is controlled by its degree. The radial basis function kernel is highly flexible and can model complex relationships by considering the distance between data points.

10. What is the effect of the C parameter in SVM

The parameter C in SVM controls the trade off between maximizing the margin and minimizing classification error. A high value of C tries to classify all training points correctly, which can lead to overfitting, while a low value allows more errors but produces a smoother decision boundary.

11. What is the role of the Gamma parameter in RBF Kernel SVM

The gamma parameter in the radial basis function kernel determines how far the influence of a single training example reaches. A high gamma focuses on nearby points and can lead to overfitting, while a low gamma considers points farther away and creates a smoother model.

12. What is the Naïve Bayes classifier, and why is it called "Naïve"?

Naive Bayes is a probabilistic classification algorithm based on Bayes theorem. It is called naive because it assumes that all features are independent of each other, which is rarely true in real data

13.  What is Bayes’ Theorem?

Bayes theorem is a mathematical formula used to calculate the probability of an event based on prior knowledge. It describes how to update the probability of a hypothesis given new evidence.

14.  Explain the differences between Gaussian Naïve Bayes, Multinomial Naïve Bayes, and Bernoulli Naïve Bayes

Gaussian Naive Bayes is used for continuous data and assumes that the data follows a normal distribution. Multinomial Naive Bayes is used for count based data such as word frequencies. Bernoulli Naive Bayes is used for binary data where features represent presence or absence.

15. When should you use Gaussian Naïve Bayes over other variants?

Gaussian Naive Bayes should be used when the features are continuous and approximately normally distributed, such as measurements or sensor data.

16. What are the key assumptions made by Naïve Bayes?

Naive Bayes makes several key assumptions including that features are independent, that each feature contributes equally to the outcome, and that the data follows a specific probability distribution depending on the variant used.

17. What are the advantages and disadvantages of Naïve Bayes?

The advantages of Naive Bayes include its simplicity, speed, and effectiveness on high dimensional data. It performs well in text classification tasks. Its disadvantages include the unrealistic independence assumption and reduced performance when features are highly correlated.

18. Why is Naïve Bayes a good choice for text classification

Naive Bayes is a good choice for text classification because text data is high dimensional and sparse. The algorithm handles word frequencies efficiently and performs well even with limited training data.

19. Compare SVM and Naïve Bayes for classification tasks

SVM and Naive Bayes differ in that SVM is a discriminative model that focuses on finding decision boundaries, while Naive Bayes is a generative model that models the probability distribution of the data. SVM generally provides higher accuracy for complex datasets but is slower, while Naive Bayes is faster and works well for simpler problems and smaller datasets.

20. How does Laplace Smoothing help in Naïve Bayes

Laplace smoothing is used in Naive Bayes to handle the problem of zero probability. It adds a small constant value to all counts so that no probability becomes zero, ensuring that unseen features do not cause the model to fail.

```python
# Write a Python program to train two SVM classifiers with Linear and RBF kernels on the Wine dataset, then compare their accuracies:
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

wine = datasets.load_wine()
X_wine = wine.data
y_wine = wine.target

X_train_wine, X_test_wine, y_train_wine, y_test_wine = train_test_split(X_wine, y_wine, test_size=0.3, random_state=42)

svm_linear = SVC(kernel='linear', random_state=42)
svm_linear.fit(X_train_wine, y_train_wine)
y_pred_linear = svm_linear.predict(X_test_wine)
accuracy_linear = accuracy_score(y_test_wine, y_pred_linear)
print(f"SVM (Linear Kernel) Accuracy on Wine dataset: {accuracy_linear:.4f}")

svm_rbf = SVC(kernel='rbf', random_state=42)
svm_rbf.fit(X_train_wine, y_train_wine)
y_pred_rbf = svm_rbf.predict(X_test_wine)
accuracy_rbf = accuracy_score(y_test_wine, y_pred_rbf)
print(f"SVM (RBF Kernel) Accuracy on Wine dataset: {accuracy_rbf:.4f}")

if accuracy_linear > accuracy_rbf:
    print("\nThe Linear Kernel performed better.")
elif accuracy_rbf > accuracy_linear:
    print("\nThe RBF Kernel performed better.")
else:
    print("\nBoth kernels performed equally well.")
```

```python
# Write a Python program to train an SVM Classifier with a Polynomial Kernel and visualize the decision boundary:
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC

X, y = make_moons(n_samples=100, noise=0.15, random_state=42)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

svm_poly = SVC(kernel='poly', degree=3, gamma='auto', random_state=42)
svm_poly.fit(X_train, y_train)

x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                     np.arange(y_min, y_max, 0.02))

Z = svm_poly.predict(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.figure(figsize=(10, 7))
plt.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.coolwarm)
plt.scatter(X[:, 0], X[:, 1], c=y, s=50, edgecolors='k', cmap=plt.cm.coolwarm)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('SVM with Polynomial Kernel Decision Boundary')
plt.show()
```

```python
#Write a Python program to train an SVM Classifier with different C values and compare the decision boundaries visually
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.svm import SVC

X, y = make_blobs(n_samples=100, centers=2, random_state=42, cluster_std=1.0)

C_values = [0.1, 1, 10, 100] # Small C (more regularization) to Large C (less regularization)

plt.figure(figsize=(15, 10))

for i, C in enumerate(C_values):
    # Initialize and train SVM with the current C value
    svm_classifier = SVC(kernel='linear', C=C, random_state=42)
    svm_classifier.fit(X, y)

    x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
    y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, 0.02),
                         np.arange(y_min, y_max, 0.02))

    Z = svm_classifier.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.subplot(2, 2, i + 1) # Arrange plots in a 2x2 grid
    plt.contourf(xx, yy, Z, alpha=0.8, cmap=plt.cm.coolwarm)
    plt.scatter(X[:, 0], X[:, 1], c=y, s=50, edgecolors='k', cmap=plt.cm.coolwarm)

    plt.scatter(svm_classifier.support_vectors_[:, 0], svm_classifier.support_vectors_[:, 1],
                s=100, facecolors='none', edgecolors='green', label='Support Vectors')

    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.title(f'SVM with C = {C}')
    plt.legend()

plt.tight_layout()
plt.show()
```

```python
#Write a Python program to train a Gaussian Naïve Bayes model and compare the predictions before and after Laplace Smoothing
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import accuracy_score

data = load_iris()
X = data.data
y = data.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=1
)
gnb = GaussianNB()
gnb.fit(X_train, y_train)
gnb_pred = gnb.predict(X_test)


scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

mnb_no_smooth = MultinomialNB(alpha=0)
mnb_no_smooth.fit(X_train_scaled, y_train)
mnb_no_pred = mnb_no_smooth.predict(X_test_scaled)

mnb_smooth = MultinomialNB(alpha=1)
mnb_smooth.fit(X_train_scaled, y_train)
mnb_pred = mnb_smooth.predict(X_test_scaled)

print("GaussianNB Accuracy:", accuracy_score(y_test, gnb_pred))
print("MultinomialNB (No Smoothing) Accuracy:", accuracy_score(y_test, mnb_no_pred))
print("MultinomialNB (Laplace Smoothing) Accuracy:", accuracy_score(y_test, mnb_pred))

print("\nSample Predictions:")
for i in range(10):
    print(f"Actual: {y_test[i]}, "
          f"GNB: {gnb_pred[i]}, "
          f"No Smooth: {mnb_no_pred[i]}, "
          f"Smoothed: {mnb_pred[i]}")
```

```python
#= Write a Python program to train an SVM Classifier and use GridSearchCV to tune the hyperparameters (C, gamma, kernel)
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

iris = datasets.load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)


param_grid = {
    'C': [0.1, 1, 10, 100],
    'gamma': [1, 0.1, 0.01, 0.001],
    'kernel': ['rbf', 'linear', 'poly']
}


grid_search = GridSearchCV(
    SVC(random_state=42),
    param_grid,
    cv=5,
    verbose=2,
    n_jobs=-1,
    scoring='accuracy'
)

print("\nStarting GridSearchCV...")
grid_search.fit(X_train, y_train)
print("GridSearchCV finished.")

print(f"\nBest parameters found: {grid_search.best_params_}")
print(f"Best cross-validation accuracy: {grid_search.best_score_:.4f}")

best_svm_classifier = grid_search.best_estimator_
y_pred_best = best_svm_classifier.predict(X_test)
accuracy_best = accuracy_score(y_test, y_pred_best)

print(f"Test set accuracy with best parameters: {accuracy_best:.4f}")
```

```python
#Write a Python program to train an SVM Classifier on an imbalanced dataset and apply class weighting and check it improve accuracy
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix


X, y = make_classification(n_samples=1000, n_features=2, n_informative=2, n_redundant=0,
                           n_clusters_per_class=1, weights=[0.9, 0.1], flip_y=0, random_state=42)

print("Dataset Class Distribution:")
unique, counts = np.unique(y, return_counts=True)
for u, c in zip(unique, counts):
    print(f"Class {u}: {c} samples")
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

print("\nTraining set Class Distribution:")
unique, counts = np.unique(y_train, return_counts=True)
for u, c in zip(unique, counts):
    print(f"Class {u}: {c} samples")

print("\nTest set Class Distribution:")
unique, counts = np.unique(y_test, return_counts=True)
for u, c in zip(unique, counts):
    print(f"Class {u}: {c} samples")

print("\n--- SVM Classifier WITHOUT Class Weighting ---")
svm_no_weights = SVC(kernel='linear', random_state=42)
svm_no_weights.fit(X_train, y_train)

y_pred_no_weights = svm_no_weights.predict(X_test)

accuracy_no_weights = accuracy_score(y_test, y_pred_no_weights)
print(f"Accuracy (No Weights): {accuracy_no_weights:.4f}")
print("Classification Report (No Weights):\n", classification_report(y_test, y_pred_no_weights))
print("Confusion Matrix (No Weights):\n", confusion_matrix(y_test, y_pred_no_weights))


print("\n--- SVM Classifier WITH Class Weighting ---")
svm_with_weights = SVC(kernel='linear', class_weight='balanced', random_state=42)
svm_with_weights.fit(X_train, y_train)

y_pred_with_weights = svm_with_weights.predict(X_test)

accuracy_with_weights = accuracy_score(y_test, y_pred_with_weights)
print(f"Accuracy (With Weights): {accuracy_with_weights:.4f}")
print("Classification Report (With Weights):\n", classification_report(y_test, y_pred_with_weights))
print("Confusion Matrix (With Weights):\n", confusion_matrix(y_test, y_pred_with_weights))

print("\n--- Comparison ---")
print(f"Accuracy (No Weights): {accuracy_no_weights:.4f}")
print(f"Accuracy (With Weights): {accuracy_with_weights:.4f}")

print("\nObserve the 'recall' and 'f1-score' for the minority class (Class 1) in both reports.")
print("Class weighting often improves the model's ability to correctly identify the minority class, \
      even if overall accuracy doesn't significantly change or slightly decreases, \
      because it prevents the model from being overly biased towards the majority class.")
```

```python
# Write a Python program to implement a Naïve Bayes classifier for spam detection using email data
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report


categories = ['rec.sport.baseball', 'sci.space', 'comp.sys.ibm.pc.hardware', 'talk.politics.mideast']

newsgroups_data = fetch_20newsgroups(subset='all', categories=categories, shuffle=True, random_state=42)

X = newsgroups_data.data  # The email content
y = newsgroups_data.target # The category labels


spam_categories_indices = [newsgroups_data.target_names.index('comp.sys.ibm.pc.hardware'),
                         newsgroups_data.target_names.index('talk.politics.mideast')]

y_binary = np.array([1 if label in spam_categories_indices else 0 for label in y])

print(f"Total samples: {len(X)}")
print(f"Spam-like samples: {np.sum(y_binary == 1)}")
print(f"Ham-like samples: {np.sum(y_binary == 0)}\n")

X_train, X_test, y_train, y_test = train_test_split(X, y_binary, test_size=0.3, random_state=42)


vectorizer = CountVectorizer(stop_words='english', lowercase=True, max_features=5000) # Limit features for efficiency

X_train_vectors = vectorizer.fit_transform(X_train)
X_test_vectors = vectorizer.transform(X_test)

print(f"Shape of X_train_vectors: {X_train_vectors.shape}")
print(f"Shape of X_test_vectors: {X_test_vectors.shape}\n")

mnb_classifier = MultinomialNB() # Alpha=1.0 by default, which is Laplace smoothing

print("Training Multinomial Naïve Bayes classifier...")
mnb_classifier.fit(X_train_vectors, y_train)
print("Training complete.\n")

y_pred = mnb_classifier.predict(X_test_vectors)

accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=['Ham', 'Spam'])

print(f"Accuracy of Naïve Bayes for Spam Detection: {accuracy:.4f}\n")
print("Classification Report:")
print(report)

print("\n--- Sample Predictions ---")
for i in range(5):
    print(f"Original text snippet: {X_test[i][:100]}...")
    print(f"Actual: {'Spam' if y_test[i] == 1 else 'Ham'}, Predicted: {'Spam' if y_pred[i] == 1 else 'Ham'}\n")
```

```python
#Write a Python program to train an SVM Classifier and a Naïve Bayes Classifier on the same dataset and compare their accuracy
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

iris = datasets.load_iris()
X = iris.data
y = iris.target

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- SVM Classifier ---
print("\n--- Training SVM Classifier ---")
svm_classifier = SVC(kernel='linear', random_state=42) # Using a linear kernel for simplicity
svm_classifier.fit(X_train, y_train)
y_pred_svm = svm_classifier.predict(X_test)
accuracy_svm = accuracy_score(y_test, y_pred_svm)
print(f"SVM Classifier Accuracy: {accuracy_svm:.4f}")

# --- Naïve Bayes Classifier (Gaussian Naïve Bayes) ---
print("\n--- Training Gaussian Naïve Bayes Classifier ---")
gnb_classifier = GaussianNB()
gnb_classifier.fit(X_train, y_train)
y_pred_gnb = gnb_classifier.predict(X_test)
accuracy_gnb = accuracy_score(y_test, y_pred_gnb)
print(f"Gaussian Naïve Bayes Classifier Accuracy: {accuracy_gnb:.4f}")

# --- Comparison ---
print("\n--- Comparison of Accuracies ---")
print(f"SVM Accuracy: {accuracy_svm:.4f}")
print(f"Gaussian Naïve Bayes Accuracy: {accuracy_gnb:.4f}")

if accuracy_svm > accuracy_gnb:
    print("\nSVM Classifier performed better on this dataset.")
elif accuracy_gnb > accuracy_svm:
    print("\nGaussian Naïve Bayes Classifier performed better on this dataset.")
else:
    print("\nBoth classifiers performed equally well on this dataset.")
```

```python
#Write a Python program to train an SVM Classifier using One-vs-Rest (OvR) and One-vs-One (OvO) strategies on the Wine dataset and compare their accuracy
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import SelectKBest, chi2

breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

print(f"Original number of features: {X_train.shape[1]}\n")

print("--- Naïve Bayes WITHOUT Feature Selection ---")
gnb_full = GaussianNB()
gnb_full.fit(X_train, y_train)
y_pred_full = gnb_full.predict(X_test)
accuracy_full = accuracy_score(y_test, y_pred_full)
print(f"Accuracy with all features: {accuracy_full:.4f}")

print("\n--- Naïve Bayes WITH Feature Selection ---")

k_features = 10
selector = SelectKBest(chi2, k=k_features)

X_train_selected = selector.fit_transform(X_train, y_train)
X_test_selected = selector.transform(X_test)

print(f"Number of features after selection: {X_train_selected.shape[1]}")

selected_feature_indices = selector.get_support(indices=True)
selected_feature_names = breast_cancer.feature_names[selected_feature_indices]
print(f"Selected features: {', '.join(selected_feature_names)}")

gnb_selected = GaussianNB()
gnb_selected.fit(X_train_selected, y_train)
y_pred_selected = gnb_selected.predict(X_test_selected)
accuracy_selected = accuracy_score(y_test, y_pred_selected)
print(f"Accuracy with selected {k_features} features: {accuracy_selected:.4f}")

print("\n--- Comparison of Accuracies ---")
print(f"Accuracy (All Features): {accuracy_full:.4f}")
print(f"Accuracy (Selected Features): {accuracy_selected:.4f}")

if accuracy_selected > accuracy_full:
    print("\nFeature selection improved the accuracy!")
elif accuracy_full > accuracy_selected:
    print("\nAccuracy was slightly better with all features.")
else:
    print("\nBoth models achieved the same accuracy.")

print("\nFeature selection can help reduce model complexity, prevent overfitting, and speed up training, even if accuracy improvements are not always dramatic.")
```

```python
#Write a Python program to train an SVM Classifier using Linear, Polynomial, and RBF kernels on the Breast Cancer dataset and compare their accuracy=
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# 1. Load the Breast Cancer dataset
breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- SVM with Linear Kernel ---
print("\n--- Training SVM with Linear Kernel ---")
svm_linear = SVC(kernel='linear', random_state=42)
svm_linear.fit(X_train, y_train)
y_pred_linear = svm_linear.predict(X_test)
accuracy_linear = accuracy_score(y_test, y_pred_linear)
print(f"SVM (Linear Kernel) Accuracy: {accuracy_linear:.4f}")

# --- SVM with Polynomial Kernel ---
print("\n--- Training SVM with Polynomial Kernel ---")
# 'degree' parameter is specific to the polynomial kernel
svm_poly = SVC(kernel='poly', degree=3, random_state=42)
svm_poly.fit(X_train, y_train)
y_pred_poly = svm_poly.predict(X_test)
accuracy_poly = accuracy_score(y_test, y_pred_poly)
print(f"SVM (Polynomial Kernel) Accuracy: {accuracy_poly:.4f}")

# --- SVM with RBF (Radial Basis Function) Kernel ---
print("\n--- Training SVM with RBF Kernel ---")
# 'gamma' parameter is specific to RBF, poly and sigmoid kernels
svm_rbf = SVC(kernel='rbf', random_state=42)
svm_rbf.fit(X_train, y_train)
y_pred_rbf = svm_rbf.predict(X_test)
accuracy_rbf = accuracy_score(y_test, y_pred_rbf)
print(f"SVM (RBF Kernel) Accuracy: {accuracy_rbf:.4f}")

# --- Comparison ---
print("\n--- Comparison of Accuracies ---")
print(f"Linear Kernel Accuracy: {accuracy_linear:.4f}")
print(f"Polynomial Kernel Accuracy: {accuracy_poly:.4f}")
print(f"RBF Kernel Accuracy: {accuracy_rbf:.4f}")

accuracies = {'Linear': accuracy_linear, 'Polynomial': accuracy_poly, 'RBF': accuracy_rbf}
best_kernel = max(accuracies, key=accuracies.get)
print(f"\nAmong the tested kernels, the {best_kernel} Kernel performed best.")
```

```python
# Write a Python program to train an SVM Classifier using Stratified K-Fold Cross-Validation and compute the average accuracy
import numpy as np
from sklearn import datasets
from sklearn.svm import SVC
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score

# 1. Load a classification dataset (e.g., Breast Cancer dataset)
breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

print(f"Dataset shape: {X.shape}")
print(f"Class distribution: {np.bincount(y)}\n")

# 2. Define the SVM Classifier
svm_classifier = SVC(kernel='linear', random_state=42)

# 3. Set up Stratified K-Fold Cross-Validation
# n_splits: number of folds
# shuffle: whether to shuffle the data before splitting into batches
# random_state: for reproducibility
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=42)

# List to store accuracy scores for each fold
accuracy_scores = []

print(f"--- Performing {n_splits}-Fold Stratified Cross-Validation ---")

# 4. Iterate through each fold
for fold, (train_index, test_index) in enumerate(skf.split(X, y)):
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]

    # Train the SVM classifier on the training data for the current fold
    svm_classifier.fit(X_train, y_train)

    # Make predictions on the test data for the current fold
    y_pred = svm_classifier.predict(X_test)

    # Calculate accuracy for the current fold
    fold_accuracy = accuracy_score(y_test, y_pred)
    accuracy_scores.append(fold_accuracy)

    print(f"Fold {fold + 1} Accuracy: {fold_accuracy:.4f}")

# 5. Compute and print the average accuracy
average_accuracy = np.mean(accuracy_scores)
std_accuracy = np.std(accuracy_scores)

print("\n--- Cross-Validation Results ---")
print(f"Individual Fold Accuracies: {np.array(accuracy_scores).round(4)}")
print(f"Average Accuracy: {average_accuracy:.4f}")
print(f"Standard Deviation of Accuracy: {std_accuracy:.4f}")

print("\nThis approach provides a more robust estimate of model performance than a single train/test split.")
```

```python
#Write a Python program to train a Naïve Bayes classifier using different prior probabilities and compare performance
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report

# 1. Load a classification dataset (e.g., Iris dataset)
iris = datasets.load_iris()
X = iris.data
y = iris.target

# Get unique classes and their original counts
unique_classes, class_counts = np.unique(y, return_counts=True)
total_samples = len(y)
original_priors = class_counts / total_samples

print(f"Original Class Distribution: {class_counts}")
print(f"Empirical Priors: {original_priors.round(4)}\n")

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# 3. Train Naïve Bayes with default (empirical) prior probabilities
print("--- Naïve Bayes with Default (Empirical) Priors ---")
gnb_default = GaussianNB()
gnb_default.fit(X_train, y_train)
y_pred_default = gnb_default.predict(X_test)

accuracy_default = accuracy_score(y_test, y_pred_default)
print(f"Accuracy (Default Priors): {accuracy_default:.4f}")
print("Classification Report (Default Priors):\n", classification_report(y_test, y_pred_default))

# 4. Train Naïve Bayes with custom (skewed) prior probabilities
# Let's skew the priors to favor class 0 significantly
# Make sure the sum of custom priors is 1.0
custom_priors = np.array([0.7, 0.15, 0.15]) # Assuming 3 classes: 0, 1, 2

# Check if custom priors sum to 1 and match number of classes
if not np.isclose(np.sum(custom_priors), 1.0):
    print("Warning: Custom priors do not sum to 1. They will be normalized internally.")
if len(custom_priors) != len(unique_classes):
    raise ValueError("Number of custom priors must match the number of classes.")

print("\n--- Naïve Bayes with Custom (Skewed) Priors ---")
print(f"Custom Priors Used: {custom_priors.round(4)}")
gnb_custom = GaussianNB(priors=custom_priors)
gnb_custom.fit(X_train, y_train)
y_pred_custom = gnb_custom.predict(X_test)

accuracy_custom = accuracy_score(y_test, y_pred_custom)
print(f"Accuracy (Custom Priors): {accuracy_custom:.4f}")
print("Classification Report (Custom Priors):\n", classification_report(y_test, y_pred_custom))

# 5. Compare performance
print("\n--- Comparison ---")
print(f"Accuracy with Default Priors: {accuracy_default:.4f}")
print(f"Accuracy with Custom Priors: {accuracy_custom:.4f}")

if accuracy_custom > accuracy_default:
    print("The classifier with custom priors performed better.")
elif accuracy_default > accuracy_custom:
    print("The classifier with default priors performed better.")
else:
    print("Both classifiers performed equally well.")

print("\nThis example shows how altering the prior probabilities can influence the model's predictions and overall performance, especially in scenarios where you have domain knowledge about class distribution or want to address class imbalance.")
```

```python
#Write a Python program to perform Recursive Feature Elimination (RFE) before training an SVM Classifier and compare accuracy
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.feature_selection import RFE
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# 1. Load a classification dataset (e.g., Breast Cancer dataset)
breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

print(f"Original number of features: {X.shape[1]}")

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# --- SVM WITHOUT RFE (Full Features) ---
print("\n--- Training SVM WITHOUT RFE ---")
svm_full = SVC(kernel='linear', random_state=42)
svm_full.fit(X_train, y_train)
y_pred_full = svm_full.predict(X_test)
accuracy_full = accuracy_score(y_test, y_pred_full)
print(f"Accuracy with all features: {accuracy_full:.4f}")

# --- SVM WITH RFE ---
print("\n--- Training SVM WITH RFE ---")

# Define the estimator (SVM) for RFE
estimator = SVC(kernel='linear', random_state=42)

# Initialize RFE
# n_features_to_select: the number of features to select (e.g., half of original features)
# step: number of features to remove at each iteration
k_features_to_select = X_train.shape[1] // 2  # Select half of the original features
rfe_selector = RFE(estimator=estimator, n_features_to_select=k_features_to_select, step=1)

# Fit RFE on the training data
rfe_selector.fit(X_train, y_train)

# Transform the training and testing data to include only selected features
X_train_selected = rfe_selector.transform(X_train)
X_test_selected = rfe_selector.transform(X_test)

print(f"Number of features after RFE: {X_train_selected.shape[1]}")

# Get the names of the selected features (if available)
selected_feature_indices = rfe_selector.get_support(indices=True)
selected_feature_names = breast_cancer.feature_names[selected_feature_indices]
print(f"Selected features by RFE: {', '.join(selected_feature_names)}")

# Train SVM with RFE selected features
svm_rfe = SVC(kernel='linear', random_state=42)
svm_rfe.fit(X_train_selected, y_train)
y_pred_rfe = svm_rfe.predict(X_test_selected)
accuracy_rfe = accuracy_score(y_test, y_pred_rfe)
print(f"Accuracy with RFE selected features: {accuracy_rfe:.4f}")

# --- Comparison ---
print("\n--- Comparison of Accuracies ---")
print(f"Accuracy (All Features): {accuracy_full:.4f}")
print(f"Accuracy (RFE Selected Features): {accuracy_rfe:.4f}")

if accuracy_rfe > accuracy_full:
    print("\nRFE improved the accuracy!")
elif accuracy_full > accuracy_rfe:
    print("\nAccuracy was slightly better with all features or RFE did not improve it.")
else:
    print("\nBoth models achieved the same accuracy.")

print("\nRFE can help in reducing model complexity, preventing overfitting, and potentially improving generalization by focusing on the most relevant features.")
```

```python
# Write a Python program to train an SVM Classifier and evaluate its performance using Precision, Recall, and F1-Score instead of accuracy
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# 1. Load a classification dataset (e.g., Breast Cancer dataset)
breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Train an SVM Classifier
print("\n--- Training SVM Classifier ---")
svm_classifier = SVC(kernel='linear', random_state=42) # Using a linear kernel for simplicity
svm_classifier.fit(X_train, y_train)

# 4. Make predictions on the test set
y_pred = svm_classifier.predict(X_test)

# 5. Evaluate performance using Precision, Recall, and F1-Score
print("\n--- Performance Evaluation ---")

# Calculate individual metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted') # 'weighted' handles class imbalance
recall = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision (weighted): {precision:.4f}")
print(f"Recall (weighted): {recall:.4f}")
print(f"F1-Score (weighted): {f1:.4f}")

# Display a full classification report for detailed per-class metrics
print("\nClassification Report:\n", classification_report(y_test, y_pred))

print("\nPrecision measures the accuracy of positive predictions.")
print("Recall measures the ability of the classifier to find all positive samples.")
print("F1-Score is the weighted harmonic mean of precision and recall.")
```

```python
#Write a Python program to train a Naïve Bayes Classifier and evaluate its performance using Log Loss (Cross-Entropy Loss)
import numpy as np
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, log_loss

# 1. Load a classification dataset (e.g., Iris dataset)
iris = datasets.load_iris()
X = iris.data
y = iris.target

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Train a Naïve Bayes Classifier (GaussianNB)
print("\n--- Training Gaussian Naïve Bayes Classifier ---")
gnb_classifier = GaussianNB()
gnb_classifier.fit(X_train, y_train)

# 4. Make predictions and predict probabilities on the test set
y_pred = gnb_classifier.predict(X_test)
y_pred_proba = gnb_classifier.predict_proba(X_test)

# 5. Evaluate performance using Accuracy and Log Loss
print("\n--- Performance Evaluation ---")

# Calculate Accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")

# Calculate Log Loss (Cross-Entropy Loss)
# Note: Log Loss requires probability estimates, not just class predictions
logloss = log_loss(y_test, y_pred_proba)
print(f"Log Loss: {logloss:.4f}")

print("\nAccuracy measures the proportion of correctly classified instances.")
print("Log Loss measures the uncertainty of the probability predictions.")
print("A lower Log Loss indicates better probability estimates.")
```

```python
# Write a Python program to train an SVM Classifier and visualize the Confusion Matrix using seaborn
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score

# 1. Load a classification dataset (e.g., Iris dataset)
iris = datasets.load_iris()
X = iris.data
y = iris.target

# 2. Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 3. Train an SVM Classifier
print("--- Training SVM Classifier ---")
svm_classifier = SVC(kernel='linear', random_state=42) # Using a linear kernel for simplicity
svm_classifier.fit(X_train, y_train)

# 4. Make predictions on the test set
y_pred = svm_classifier.predict(X_test)

# 5. Evaluate overall accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy: {accuracy:.4f}")

# 6. Generate the Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

# 7. Visualize the Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for SVM Classifier')
plt.show()

print("\nThe Confusion Matrix visually represents the performance of a classification algorithm.")
print("Rows correspond to true labels, and columns correspond to predicted labels.")
print("The diagonal elements show the number of correct predictions for each class.")
print("Off-diagonal elements indicate misclassifications.")
```

```python
#Write a Python program to train an SVM Classifier and visualize the Precision-Recall Curve
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import precision_recall_curve, auc, average_precision_score

# 1. Load a classification dataset (e.g., Breast Cancer dataset for binary classification)
breast_cancer = datasets.load_breast_cancer()
X = breast_cancer.data
y = breast_cancer.target

# 2. Split the dataset into training and testing sets
# Use 'stratify=y' to maintain the same class distribution in train and test sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# 3. Train an SVM Classifier
# For precision_recall_curve, we need probability estimates or decision function scores.
# SVC with probability=True enables predict_proba, but it's computationally expensive.
# Alternatively, use decision_function for scores for linear kernels or for non-linear
# kernels where decision_function returns distance to hyperplane.
print("--- Training SVM Classifier ---")
svm_classifier = SVC(kernel='linear', random_state=42, probability=True) # Set probability=True
svm_classifier.fit(X_train, y_train)

# 4. Get prediction probabilities for the positive class
# For binary classification, y_pred_proba[:, 1] gives probabilities of the positive class
y_pred_proba = svm_classifier.predict_proba(X_test)[:, 1]

# 5. Compute Precision-Recall curve and AUC
precision, recall, _ = precision_recall_curve(y_test, y_pred_proba)
# Calculate Area Under the Precision-Recall Curve (AUPRC)
auprc = auc(recall, precision)
# Calculate Average Precision Score (equivalent to AUPRC for binary classification)
avg_precision = average_precision_score(y_test, y_pred_proba)

print(f"Average Precision Score (AUPRC): {avg_precision:.4f}")

# 6. Plot the Precision-Recall Curve
plt.figure(figsize=(8, 6))
plt.plot(recall, precision, color='blue', lw=2, label=f'PR curve (area = {auprc:.2f})')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve for SVM Classifier')
plt.legend(loc='lower left')
plt.grid(True)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.show()

print("\nThe Precision-Recall curve shows the trade-off between precision and recall for different threshold settings.")
print("A higher AUPRC indicates better performance, especially for imbalanced datasets.")
```
